## Data Cleaning
Strip whitespace from column names and rename the misspelled `sectoer` column to `sector`.

```python
df.columns = df.columns.str.strip()
df = df.rename(columns={'sectoer':'sector'})
```

Confirm the column names are now clean.

```python
df.columns
```

Convert `date` to a proper datetime type and `no_trades` from float to integer, since trade counts are always whole numbers.

```python
# 1. Convert the 'date' column to datetime format
df['date'] = pd.to_datetime(df['date'])

# 2. Convert 'no_trades' from float to integer
df['no_trades'] = df['no_trades'].astype(int)
```

Confirm the dtype fixes took effect.

```python
df.dtypes
```

Recheck for missing values now that columns are cleaned — `open`, `high`, and `low` still have 162 missing values each.

```python
df.isna().sum()
```

### Handle Missing Values
Inspect the rows where `open`, `high`, and `low` are all missing to understand the pattern before deciding how to fill them.

```python
nulls = df[(df['open'].isnull()) & (df['high'].isnull()) & (df['low'].isnull())]
nulls
```

Break down which trading names these missing rows belong to.

```python
nulls['trading_name'].value_counts()
```

Impute the missing `open`, `high`, and `low` values with each column's mean.

```python
cols = ['open','high','low']

for col in cols:
    df[col] = df[col].fillna(df[col].mean())
```

Verify no rows remain with `open`, `high`, and `low` all missing.

```python
df[(df['open'].isnull()) & (df['high'].isnull()) & (df['low'].isnull())]
```

Final check: confirm there are no missing values left in any column.

```python
df.isna().sum()
```
